# Tornado-model visualizations

All calculations and plotting logic are imported from `mod_tornado_functions.py`. The notebook has no local function definitions and does not use a settings dataclass.

The vorticity plots use
$\zeta(r)=\Gamma a(4\pi\nu_v)^{-1}\exp[-ar^2/(4\nu_v)]$,
$u=-ar/2$, $w=az$, and the consistent circulation-derived azimuthal wind.
The wind plots compare the corrected axisymmetric azimuthal-momentum solutions for $w=az+br$ and $w=az$.
